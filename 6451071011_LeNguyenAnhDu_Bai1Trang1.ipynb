# Bài 1: Phân lớp bộ dữ liệu Iris

**Họ tên:** Lê Nguyễn Anh Du  
**MSSV:** 6451071011

Thực hiện đầy đủ các yêu cầu a–j bằng scikit-learn và TensorFlow/Keras.

```python
print("Họ tên: Lê Nguyễn Anh Du | MSSV: 6451071011")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import tensorflow as tf
HO_TEN, MSSV = "Lê Nguyễn Anh Du", "6451071011"
def in_thong_tin(): print(f"Họ tên: {HO_TEN} | MSSV: {MSSV}")
np.random.seed(42)
tf.random.set_seed(42)
in_thong_tin()
print("Đã nhập thư viện và thiết lập seed.")
```

## a) Đọc dữ liệu Iris từ sklearn

```python
in_thong_tin()
iris = load_iris()
X, y = iris.data, iris.target
print("Đã tải bộ dữ liệu Iris.")
```

## b) Khảo sát bộ dữ liệu

```python
in_thong_tin()
print("Tên thuộc tính:", iris.feature_names)
print("Tên lớp:", iris.target_names)
print("Số lượng mẫu:", X.shape[0])
print("Số lượng thuộc tính:", X.shape[1])
print("Kích thước target:", y.shape)
print("\n5 mẫu đầu tiên (thuộc tính):\n", X[:5])
print("Nhãn 5 mẫu đầu tiên:", y[:5])
```

```python
in_thong_tin()
df_iris = pd.DataFrame(X, columns=iris.feature_names)
print("Thống kê count, mean, std, min, max và các tứ vị phân:")
display(df_iris.describe())
```

## c) Chuẩn hóa dữ liệu về đoạn [0, 1]

Chia từng giá trị cho giá trị lớn nhất trong cột thuộc tính tương ứng.

```python
in_thong_tin()
max_moi_cot = X.max(axis=0)
X_scaled = X / max_moi_cot
print("Giá trị lớn nhất mỗi thuộc tính:", max_moi_cot)
print("5 mẫu sau chuẩn hóa:\n", X_scaled[:5])
```

## d) Chia dữ liệu thành tập train và test theo tỷ lệ 8:2

```python
in_thong_tin()
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)
print("X_train:", X_train.shape, "| X_test:", X_test.shape)
print("y_train:", y_train.shape, "| y_test:", y_test.shape)
```

## Hàm đánh giá mô hình

In accuracy, precision, recall, F1-score theo từng lớp và trung bình; đồng thời vẽ confusion matrix bằng heatmap.

```python
def danh_gia_mo_hinh(ten, y_true, y_pred):
    in_thong_tin()
    print(f"\n===== {ten} =====")
    print(f"Accuracy: {accuracy_score(y_true, y_pred):.4f}")
    print(classification_report(y_true, y_pred, target_names=iris.target_names, digits=4, zero_division=0))
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=iris.target_names, yticklabels=iris.target_names)
    plt.xlabel("Nhãn dự đoán")
    plt.ylabel("Nhãn thực tế")
    plt.title(f"Confusion matrix - {ten}")
    plt.tight_layout()
    plt.show()
```

## e) Phân lớp bằng Cây quyết định (Decision Tree)

```python
in_thong_tin()
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)
print("Đã huấn luyện Decision Tree và dự đoán trên tập test.")
danh_gia_mo_hinh("Decision Tree", y_test, dt_pred)
```

## f) Phân lớp bằng Naïve Bayes

```python
in_thong_tin()
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
nb_pred = nb_model.predict(X_test)
print("Đã huấn luyện Gaussian Naïve Bayes và dự đoán trên tập test.")
danh_gia_mo_hinh("Gaussian Naïve Bayes", y_test, nb_pred)
```

## g) Phân lớp bằng K Nearest Neighbors (KNN)

```python
in_thong_tin()
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
knn_pred = knn_model.predict(X_test)
print("Đã huấn luyện KNN (k=5) và dự đoán trên tập test.")
danh_gia_mo_hinh("K Nearest Neighbors (k=5)", y_test, knn_pred)
```

## h) Phân lớp bằng Neural Network

Mạng gồm hai hidden layer: lớp thứ nhất có 10 neuron, lớp thứ hai có 20 neuron (ReLU); output layer có 3 neuron (Softmax). Huấn luyện 100 epoch, batch size bằng 1.

```python
in_thong_tin()
nn_model = Sequential([
    Dense(10, activation="relu", input_shape=(X_train.shape[1],)),
    Dense(20, activation="relu"),
    Dense(len(iris.target_names), activation="softmax")
])
nn_model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
history = nn_model.fit(X_train, y_train, epochs=100, batch_size=1, verbose=1)
```

```python
in_thong_tin()
nn_prob = nn_model.predict(X_test, verbose=0)
nn_pred = np.argmax(nn_prob, axis=1)
print("Nhãn dự đoán được chọn bằng argmax trên xác suất mỗi lớp.")
danh_gia_mo_hinh("Neural Network", y_test, nn_pred)
```

## i) Dự đoán ba mẫu dữ liệu mới

Chuẩn hóa các mẫu bằng cùng giá trị lớn nhất mỗi thuộc tính đã dùng cho dữ liệu Iris.

```python
in_thong_tin()
mau_moi = np.array([[6.2, 2.9, 4.3, 1.3], [5.1, 3.5, 1.4, 0.2], [7.3, 2.8, 6.4, 2.1]])
mau_moi_scaled = mau_moi / max_moi_cot
for ten, model in {"Decision Tree": dt_model, "Naïve Bayes": nb_model, "KNN (k=5)": knn_model}.items():
    print(f"{ten}: {iris.target_names[model.predict(mau_moi_scaled)]}")
nn_moi = np.argmax(nn_model.predict(mau_moi_scaled, verbose=0), axis=1)
print(f"Neural Network: {iris.target_names[nn_moi]}")
```

## j) Nhận xét, đánh giá hiệu suất

So sánh các mô hình trên cùng tập test bằng accuracy, precision, recall, F1-score. Đối chiếu confusion matrix để nhận biết lớp nào được phân loại đúng hoặc bị nhầm lẫn. Bảng dưới đây xếp hạng accuracy trong lần chia dữ liệu này.

```python
in_thong_tin()
kq = {"Decision Tree": accuracy_score(y_test, dt_pred), "Naïve Bayes": accuracy_score(y_test, nb_pred), "KNN (k=5)": accuracy_score(y_test, knn_pred), "Neural Network": accuracy_score(y_test, nn_pred)}
bang_kq = pd.DataFrame([{"Mô hình": ten, "Accuracy": acc} for ten, acc in kq.items()]).sort_values("Accuracy", ascending=False).reset_index(drop=True)
display(bang_kq)
print(f"Mô hình có accuracy cao nhất: {bang_kq.loc[0, 'Mô hình']} ({bang_kq.loc[0, 'Accuracy']:.4f}).")
print("Dựa vào báo cáo và biểu đồ phía trên, nhận xét ưu điểm, hạn chế và các lớp được phân loại đúng hoặc nhầm lẫn.")
```
